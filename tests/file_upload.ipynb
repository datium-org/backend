```python
import os
from db import S3Client

# Example: Initialize S3Client with AWS S3
# s3 = S3Client(bucket_name='your-s3-bucket-name')

# Example: Initialize S3Client with MinIO (Local Development)
s3 = S3Client(
  bucket_name='test-bucket',
  endpoint_url='http://localhost:9000',
  aws_access_key_id='minioadmin',
  aws_secret_access_key='minioadmin',
  aws_region='us-east-1',
  use_ssl=False,
  verify_ssl=False
)
```

```python
success = s3.upload_file(
    file_path='sample_docs/MUD LAB QUOTE.pdf',
    object_name='documents/file.txt',
)

if success:
    print("Upload successful!")
else:
    print("Upload failed.")
```

```python
success = s3.delete_file(object_name='documents/file.txt')
if success:
    print("File deleted successfully!")
else:
    print("Failed to delete file.")
```

```python
metadata = s3.get_object_metadata(object_name='documents/file.txt')
if metadata:
    print("Object Metadata:", metadata)
else:
    print("Failed to retrieve metadata.")
```
