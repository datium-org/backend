```python
import nest_asyncio

nest_asyncio.apply()

from llama_parse import LlamaParse
import os
from dotenv import load_dotenv

load_dotenv("../.env")
api_key = os.getenv("LLAMA_INDEX_API_KEY")  # get from cloud.llamaindex.ai
```

```python
parser = LlamaParse(
    api_key=api_key,  # can also be set in your env as LLAMA_CLOUD_API_KEY
    result_type="markdown",
)

documents = parser.load_data("sample_docs/Oilfield Production Chemicals.xlsx")
```

```python
documents
```

```python


with open("../.env", "r+") as f:
  print(f.read())

from llama_index.llms.openai import OpenAI
from llama_index.core import Settings

llm = OpenAI(model="gpt-4o")
Settings.llm = llm
```

```python
from llama_index.core import VectorStoreIndex

index = VectorStoreIndex.from_documents(documents)

query_engine = index.as_query_engine()
```

```python
response = query_engine.query("List all the cementing chemicals which have a cost of usage lower than 100,000")
print(str(response))
```
